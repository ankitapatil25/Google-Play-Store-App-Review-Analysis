## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Remove duplicate rows from Play Store dataset

df_ps = df_ps.drop_duplicates()
print("Duplicates removed successfully.")
````

````python
# Convert Installs column to string first
df_ps['Installs'] = df_ps['Installs'].astype(str)

# clean the column
df_ps['Installs'] = df_ps['Installs'].str.replace('+', '', regex=False)
df_ps['Installs'] = df_ps['Installs'].str.replace(',', '', regex=False)

# Convert to numeric
df_ps['Installs'] = pd.to_numeric(df_ps['Installs'], errors='coerce')

# Drop missing values (critical column)
df_ps = df_ps.dropna(subset=['Installs'])

print("Installs column cleaned and missing values handled.")
````

````python
# Convert to string first
df_ps['Price'] = df_ps['Price'].astype(str)

# Remove '$' symbol
df_ps['Price'] = df_ps['Price'].str.replace('$', '', regex=False)

# Convert to numeric
df_ps['Price'] = pd.to_numeric(df_ps['Price'], errors='coerce')

# Fill missing values WITHOUT inplace (FIXED)
df_ps['Price'] = df_ps['Price'].fillna(0)

print("Price column cleaned and missing values handled.")
````

````python
# Reload ONLY Size column from original dataset
df_ps_raw = pd.read_csv(ps_data_path)

# Replace Size column with fresh data
df_ps['Size'] = df_ps_raw['Size']


# Convert to string
df_ps['Size'] = df_ps['Size'].astype(str)

# Remove unwanted text
df_ps['Size'] = df_ps['Size'].str.replace('Varies with device', '')

# Remove commas
df_ps['Size'] = df_ps['Size'].str.replace(',', '')

# Convert values
df_ps['Size'] = df_ps['Size'].apply(lambda x:
    float(x[:-1]) if 'M' in x else
    (float(x[:-1]) / 1024 if 'k' in x else np.nan)
)

# Convert to numeric
df_ps['Size'] = pd.to_numeric(df_ps['Size'], errors='coerce')

# Check nulls before filling
print("Null before:", df_ps['Size'].isnull().sum())

# Fill with median
df_ps['Size'] = df_ps['Size'].fillna(df_ps['Size'].median())

# Final check
print("Null after:", df_ps['Size'].isnull().sum())
print("Datatype:", df_ps['Size'].dtype)
````

````python
# Convert 'Reviews' column to numeric

df_ps['Reviews'] = pd.to_numeric(df_ps['Reviews'], errors='coerce')

# Fill missing Reviews with median
df_ps['Reviews'].fillna(df_ps['Reviews'].median())

print("Reviews column cleaned and missing values handled.")
````

````python
# Check central tendency measures for Rating

print("Mean:", df_ps['Rating'].mean())
print("Median:", df_ps['Rating'].median())
print("Mode:", df_ps['Rating'].mode()[0])
````

````python
# Convert 'Rating' column to numeric
df_ps['Rating'] = pd.to_numeric(df_ps['Rating'], errors='coerce')

# Fill missing values using median (robust choice)
df_ps['Rating'] = df_ps['Rating'].fillna(df_ps['Rating'].median())

print("Rating column cleaned and missing values handled using median.")
````

````python
#check the last 5 rows of the dataframe after data cleaning
df_reviews.tail()
````

````python
#check if any null vale in dataframe
df_reviews.isna().sum()
````

````python
# Count number of sentiments (Positive, Negative, Neutral) per App

# Group by App and Sentiment, then count occurrences
g_df = df_reviews.groupby(['App', 'Sentiment']).size().unstack()

# Replace missing values with 0 (apps may not have all sentiment types)
g_df = g_df.fillna(0)

# Reset index to make 'App' a column again
g_df = g_df.reset_index()

# Rename columns for clarity
g_df.columns = ['App', 'Negative_Sentiment', 'Neutral_Sentiment', 'Positive_Sentiment']

# Display first few rows
g_df.head()
````

````python
# Calculate average sentiment polarity and subjectivity per App

# Group by App and calculate mean
sentiment_mean = df_reviews.groupby('App').agg({
    'Sentiment_Polarity': 'mean',
    'Sentiment_Subjectivity': 'mean'
}).reset_index()

# Rename columns for better understanding
sentiment_mean.rename(columns={
    'Sentiment_Polarity': 'Mean_Sentiment_Polarity',
    'Sentiment_Subjectivity': 'Mean_Sentiment_Subjectivity'
}, inplace=True)

# Display result
sentiment_mean.head()
````

````python
# Combine sentiment count data with mean sentiment data

# Merge both sentiment datasets on 'App'
g_df = pd.merge(g_df, sentiment_mean, on='App', how='inner')

# Display merged sentiment dataframe
g_df.head()
````

````python

# Check for missing values after merging

g_df.isnull().sum()
````

````python
# Merge Play Store dataset with sentiment dataset

merged_df = pd.merge(df_ps, g_df, on='App', how='inner')

# Display first few rows
merged_df.head()
````

````python
# Check shape of final merged dataset

print("Shape of merged dataset:", merged_df.shape)
````
